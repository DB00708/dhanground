# 07 ROI

Measures the A3 arm's answer throughput and token counts on dev rows, then compares self-hosted cost per 1M
answers (GPU price per hour) against a closed model at the vendor's published per-token prices.
Output: `evals/results/roi.json`. Manifest: none; `roi.json` stores every input price it used.

**Needs:** a GPU (L4) for `--measure`. Set `P_IN` / `P_OUT` to the vendor's published USD per 1M input/output
tokens on the day, and cite the pricing page in the write-up.

```python
import json, os, sys
from collections import Counter
from pathlib import Path

# Run from the repo root so relative paths match the README runbook.
ROOT = Path.cwd()
while not (ROOT / "core").is_dir() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
os.chdir(ROOT)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
PY = sys.executable  # `!{PY} -m ...` runs the CLI with this kernel's interpreter
print("repo root:", ROOT)
```

```python
ARM = "A3"
P_IN = None   # USD per 1M input tokens
P_OUT = None  # USD per 1M output tokens
assert P_IN is not None and P_OUT is not None, "set P_IN and P_OUT from the vendor's pricing page"
```

```python
!{PY} -m evals.roi --measure --arm {ARM} --p-in {P_IN} --p-out {P_OUT}
```
